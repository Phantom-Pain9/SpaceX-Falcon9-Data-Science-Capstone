In this lab, we perform exploratory data analysis and create our binary target variable `Class` indicating landing success.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/dataset_part_2.csv')
print("Dataset shape:", df.shape)
df.head()
```

### Task 1: Calculate Launches at Each Site

```python
launch_site_counts = df['LaunchSite'].value_counts()
print("Launch Site Counts:\n", launch_site_counts)
```

### Task 2: Calculate Number and Occurrence of Each Orbit

```python
orbit_counts = df['Orbit'].value_counts()
print("Orbit Counts:\n", orbit_counts)
```

### Task 3: Calculate Landing Outcomes & Create Target Variable `Class`

```python
landing_outcomes = df['Outcome'].value_counts()
print("Landing Outcomes:\n", landing_outcomes)

# Successful landing: True Ocean, True RTLS, True ASDS
# Unsuccessful landing: False Ocean, False RTLS, False ASDS, None ASDS, None None
print("\nClass distribution (1: Success, 0: Failure):")
print(df['Class'].value_counts())
print(f"\nOverall Landing Success Rate: {df['Class'].mean() * 100:.2f}%")
```
