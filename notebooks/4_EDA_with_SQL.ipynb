# Exploratory Data Analysis (EDA) with SQL
### IBM Data Science Professional Certificate - Capstone Project
Executing SQL queries against the SpaceX database `SPACEXTBL` to analyze missions and landing records.

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('../data/my_data1.db')
print("Connected to SQLite database.")
```

### Task 1: Unique Launch Sites

```python
pd.read_sql("SELECT DISTINCT Launch_Site FROM SPACEXTBL;", conn)
```

### Task 2: Launch Sites Beginning with 'CCA'

```python
pd.read_sql("SELECT Launch_Site FROM SPACEXTBL WHERE Launch_Site LIKE 'CCA%' LIMIT 5;", conn)
```

### Task 3: Total Payload Mass Carried by NASA (CRS)

```python
pd.read_sql("SELECT SUM(PAYLOAD_MASS__KG_) AS Total_Payload_Mass_KG FROM SPACEXTBL WHERE Customer = 'NASA (CRS)';", conn)
```

### Task 4: Average Payload Mass for Booster Version F9 v1.1

```python
pd.read_sql("SELECT AVG(PAYLOAD_MASS__KG_) AS Average_Payload_Mass_KG FROM SPACEXTBL WHERE Booster_Version LIKE 'F9 v1.1%';", conn)
```

### Task 5: Date of First Successful Ground Pad Landing

```python
pd.read_sql("SELECT MIN(Date) AS First_Successful_Ground_Pad_Landing_Date FROM SPACEXTBL WHERE Landing_Outcome = 'Success (ground pad)';", conn)
```

### Task 6: Drone Ship Boosters with Payload between 4000 and 6000 kg

```python
pd.read_sql("SELECT Booster_Version, PAYLOAD_MASS__KG_ FROM SPACEXTBL WHERE Landing_Outcome = 'Success (drone ship)' AND PAYLOAD_MASS__KG_ BETWEEN 4000 AND 6000;", conn)
```

### Task 7: Total Successful and Failure Mission Outcomes

```python
pd.read_sql("SELECT Mission_Outcome, COUNT(*) AS Total_Count FROM SPACEXTBL GROUP BY Mission_Outcome;", conn)
```

### Task 8: Boosters Carrying Maximum Payload Mass

```python
pd.read_sql("SELECT Booster_Version, PAYLOAD_MASS__KG_ FROM SPACEXTBL WHERE PAYLOAD_MASS__KG_ = (SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTBL);", conn)
```

### Task 9: Drone Ship Failures in 2015

```python
pd.read_sql("SELECT Booster_Version, Launch_Site, Landing_Outcome, Date FROM SPACEXTBL WHERE Landing_Outcome = 'Failure (drone ship)' AND Date LIKE '%2015%';", conn)
```

### Task 10: Rank Count of Landing Outcomes (2010-06-04 to 2017-03-20)

```python
pd.read_sql("SELECT Landing_Outcome, COUNT(*) AS Outcome_Count FROM SPACEXTBL WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Landing_Outcome ORDER BY Outcome_Count DESC;", conn)
```
