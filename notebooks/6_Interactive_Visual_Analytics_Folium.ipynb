# Interactive Visual Analytics with Folium
### IBM Data Science Professional Certificate - Capstone Project
Geospatial analysis of SpaceX launch sites, landing outcomes, and proximity to infrastructure.

```python
import folium
import pandas as pd
from folium.plugins import MarkerCluster

spacex_df = pd.read_csv('../data/spacex_launch_geo.csv')
print("Loaded spacex_launch_geo.csv shape:", spacex_df.shape)
spacex_df.head()
```

### Task 1: Mark Launch Sites on Folium Map

```python
site_map = folium.Map(location=[28.562302, -80.577356], zoom_start=4)
for _, row in spacex_df[['Launch Site', 'Lat', 'Long']].drop_duplicates().iterrows():
    circle = folium.Circle([row['Lat'], row['Long']], radius=1000, color='#d35400', fill=True)
    marker = folium.map.Marker([row['Lat'], row['Long']], icon=folium.DivIcon(icon_size=(20,20), icon_anchor=(0,0), html=f'<div style="font-size: 12; color:#d35400;"><b>{row["Launch Site"]}</b></div>'))
    site_map.add_child(circle)
    site_map.add_child(marker)
site_map
```

### Task 2: MarkerClusters for Success & Failure Launches

```python
marker_cluster = MarkerCluster().add_to(site_map)
for _, row in spacex_df.iterrows():
    marker_color = 'green' if row['class'] == 1 else 'red'
    marker = folium.Marker([row['Lat'], row['Long']], icon=folium.Icon(color='white', icon_color=marker_color))
    marker_cluster.add_child(marker)
site_map
```
